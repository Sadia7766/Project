```python
import nltk
import random

nltk.download('punkt')

# Define patterns and responses
patterns = {
    "hi": ["Hello!", "Hi there!", "Hey!"],
    "how are you": ["I'm doing well, thanks for asking.", "Doing great, how about you?", "Fantastic, and you?"],
    "fine,thank you":["Welcome.","It's my pleasure.","Welcome,happy to hear that."],
    "what is your name": ["My name is Chatbot. It's nice to meet you!", "I'm Chatbot, a pleasure to meet you.", "I'm the Chatbot, at your service."],
    "bye": ["Goodbye!", "See you later!", "Have a great day!"],
    "thank you": ["You're welcome!", "No problem!", "Glad I could help."]
}


def chat():
    print("Start chatting with the bot (type 'quit' to stop)!")
    while True:
        user_input = input("You: ").lower()
        if user_input == "quit":
            break

        response = "I'm sorry, I didn't understand that. Could you please rephrase your question?"
        for pattern, responses in patterns.items():
            if pattern in user_input:
                response = random.choice(responses)
                break

        print("Chatbot: ", response)

chat()
```
